```python
import requests
from io import BytesIO
from PIL import Image
import matplotlib.pyplot as plt

# 이미지 URL 설정
url = 'https://encrypted-tbn0.gstatic.com/images?q=tbn:ANd9GcS_mkaCqysyuQo-dzVebBc4beXoWIRj0FDrrvk7-j8gRQ&s=10'

try:
    # URL로부터 이미지 데이터를 가져옴
    response = requests.get(url)
    response.raise_for_status()

    # 바이너리 I/O를 사용하여 이미지를 오픈
    image_bytes = BytesIO(response.content)
    img = Image.open(image_bytes)

    # matplotlib을 이용해 시각화
    plt.figure(figsize=(5, 5))
    plt.imshow(img)
    plt.axis('off')  # 축 정보 숨기기
    plt.show()
except Exception as e:
    print(f"이미지를 불러오는 중 오류가 발생했습니다: {e}")
```

```python
import random
import sqlite3
from faker import Faker

# 러시아어 로케일 설정
fake = Faker('ru_RU')
db_filename = 'shop.db'

# 1. 데이터베이스 연결 (외래 키 활성화 필수)
conn = sqlite3.connect(db_filename)
cursor = conn.cursor()

# SQLite는 기본적으로 FK 제약조건이 꺼져 있으므로 활성화해야 합니다.
cursor.execute("PRAGMA foreign_keys = ON;")

# 2. 테이블 생성 (1:N 관계)
# 사용자 테이블
cursor.execute('''
CREATE TABLE IF NOT EXISTS users (
    id INTEGER PRIMARY KEY AUTOINCREMENT,
    name TEXT NOT NULL,
    email TEXT UNIQUE NOT NULL
)
''')

# 주문 테이블 (user_id가 users(id)를 참조)
cursor.execute('''
CREATE TABLE IF NOT EXISTS orders (
    id INTEGER PRIMARY KEY AUTOINCREMENT,
    user_id INTEGER,
    product_name TEXT NOT NULL,
    price INTEGER NOT NULL,
    order_date TEXT,
    FOREIGN KEY (user_id) REFERENCES users (id) ON DELETE CASCADE
)
''')
conn.commit()


print("=== 1. Faker로 연관 데이터 생성 및 적재 ===")

# 기존 데이터 초기화 (재실행 시 중복 방지)
cursor.execute("DELETE FROM orders")
cursor.execute("DELETE FROM users")

# 사용자 5명 생성
users_data = []
for _ in range(5):
    users_data.append((fake.name(), fake.email()))

cursor.executemany('''
INSERT INTO users (name, email) VALUES (?, ?)
''', users_data)
conn.commit()

# 각 사용자별로 1~3개의 무작위 주문 생성
products = ['Смартфон', 'Ноутбук', 'Наушники', 'Планшет', 'Клавиатура', 'Монитор']
cursor.execute("SELECT id FROM users")
user_ids = [row[0] for row in cursor.fetchall()]

orders_data = []
for uid in user_ids:
    num_orders = random.randint(1, 3) # 사용자당 1~3개 주문
    for _ in range(num_orders):
        orders_data.append((
            uid,
            random.choice(products),
            random.randint(5000, 150000), # 가격 (루블 단위 가정)
            fake.date_this_year().strftime('%Y-%m-%d')
        ))

cursor.executemany('''
INSERT INTO orders (user_id, product_name, price, order_date)
VALUES (?, ?, ?, ?)
''', orders_data)
conn.commit()
print("데이터 적재 완료!\n")


# ==========================================
# 3. 고급 쿼리 실습 (JOIN 및 집계)
# ==========================================

print("--- [JOIN 조회] 고객별 주문 내역 확인 ---")
# INNER JOIN을 사용해 사용자와 주문 정보를 함께 조회
cursor.execute('''
SELECT users.name, orders.product_name, orders.price, orders.order_date
FROM orders
JOIN users ON orders.user_id = users.id
ORDER BY orders.order_date DESC
''')

for row in cursor.fetchall():
    print(f"고객명: {row[0]} | 상품: {row[1]} | 가격: {row[2]}₽ | 일자: {row[3]}")
print()


print("--- [집계 쿼리] 고객별 총 주문 금액 및 주문 횟수 ---")
# GROUP BY와 집계 함수(SUM, COUNT) 활용
cursor.execute('''
SELECT users.name, COUNT(orders.id) AS total_orders, SUM(orders.price) AS total_spent
FROM users
LEFT JOIN orders ON users.id = orders.user_id
GROUP BY users.id
''')

for row in cursor.fetchall():
    print(f"고객명: {row[0]} | 총 주문 횟수: {row[1]}회 | 총 지출: {row[2] or 0}₽")


# 4. 연결 종료
conn.close()
print("\n=== 실습 완료 및 DB 연결 종료 ===")
```

```python
import csv
import sqlite3
from faker import Faker

def generate_fake_csv(filename="users.csv", num_rows=5):
    """1. Faker를 이용해 가상의 사용자 CSV 파일을 생성합니다."""
    # 한국어 로케일 설정 (영어 데이터를 원하면 'en_US'로 변경 가능)
    fake = Faker('ko_KR')

    with open(filename, mode='w', encoding='utf-8', newline='') as f:
        writer = csv.writer(f)
        # 헤더 작성
        writer.writerow(['name', 'email', 'address', 'job'])

        for _ in range(num_rows):
            writer.writerow([
                fake.name(),
                fake.email(),
                fake.address().replace('\n', ' '), # 줄바꿈 제거
                fake.job()
            ])
    print(f"[+] '{filename}' 파일 생성 완료 ({num_rows}행)")

def main():
    csv_filename = "users.csv"
    db_filename = "my_database.db"

    # Step 1: 가상 CSV 파일 생성
    generate_fake_csv(csv_filename, num_rows=5)

    # SQLite 데이터베이스 연결 (파일이 없으면 자동으로 생성됩니다)
    conn = sqlite3.connect(db_filename)
    cursor = conn.cursor()

    # 테이블 생성
    cursor.execute('''
        CREATE TABLE IF NOT EXISTS users (
            id INTEGER PRIMARY KEY AUTOINCREMENT,
            name TEXT NOT NULL,
            email TEXT UNIQUE,
            address TEXT,
            job TEXT
        )
    ''')
    conn.commit()

    # Step 2: CSV 데이터를 읽어서 SQLite DB에 적재 (INSERT)
    with open(csv_filename, mode='r', encoding='utf-8') as f:
        reader = csv.reader(f)
        next(reader) # 첫 줄(헤더) 건너뛰기
        for row in reader:
            try:
                cursor.execute('''
                    INSERT OR IGNORE INTO users (name, email, address, job)
                    VALUES (?, ?, ?, ?)
                ''', row)
            except sqlite3.Error as e:
                print(f"데이터 삽입 중 오류 발생: {e}")
        conn.commit()
    print("[+] CSV 데이터를 SQLite DB에 적재 완료\n")

    # ==========================================
    # Step 3: CRUD 실습
    # ==========================================

    # 1. READ (전체 조회)
    print("--- [READ] 전체 사용자 목록 조회 ---")
    cursor.execute("SELECT id, name, email, job FROM users")
    for row in cursor.fetchall():
        print(row)

    # 2. CREATE (새로운 데이터 추가)
    print("\n--- [CREATE] 새로운 사용자 추가 ---")
    new_user = ("홍길동", "hong@example.com", "서울시 강남구 테헤란로", "소프트웨어 엔지니어")
    cursor.execute('''
        INSERT INTO users (name, email, address, job)
        VALUES (?, ?, ?, ?)
    ''', new_user)
    conn.commit()
    print("[+] '홍길동' 추가 완료")

    # 3. READ (조건 조회)
    print("\n--- [READ] '홍길동' 정보 조회 ---")
    cursor.execute("SELECT * FROM users WHERE name = ?", ("홍길동",))
    print(cursor.fetchone())

    # 4. UPDATE (데이터 수정)
    print("\n--- [UPDATE] '홍길동' 직업 정보 수정 ---")
    cursor.execute('''
        UPDATE users SET job = ? WHERE name = ?
    ''', ("시니어 백엔드 개발자", "홍길동"))
    conn.commit()

    # 수정 확인
    cursor.execute("SELECT name, job FROM users WHERE name = ?", ("홍길동",))
    print(f"수정 결과: {cursor.fetchone()}")

    # 5. DELETE (데이터 삭제)
    print("\n--- [DELETE] '홍길동' 정보 삭제 ---")
    cursor.execute("DELETE FROM users WHERE name = ?", ("홍길동",))
    conn.commit()
    print("[+] '홍길동' 삭제 완료")

    # 최종 데이터 개수 확인
    cursor.execute("SELECT COUNT(*) FROM users")
    print(f"\n[+] 현재 DB에 남아있는 총 사용자 수: {cursor.fetchone()[0]}명")

    # 데이터베이스 연결 종료
    conn.close()

if __name__ == "__main__":
    main()
```
