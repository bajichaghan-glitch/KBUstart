```python
# @title **3개 파일을 정리하여 데이터를 파악하는 코드**

def summarize_dataframe(df, df_name):
    print(f"\n--- Summary for DataFrame: {df_name} ---")
    summary = pd.DataFrame(index=df.columns)
    summary["Data Type"] = df.dtypes
    summary["Unique Values"] = df.nunique()
    summary["Missing Values (%)"] = df.isnull().sum() * 100 / len(df)

    top_values_list = []
    for col in df.columns:
        if df[col].nunique() > 0:
            top_5_values = df[col].value_counts().nlargest(5).to_dict()
            top_values_list.append(str(top_5_values))
        else:
            top_values_list.append("N/A")

    summary["Top 5 Values"] = top_values_list

    # Convert float columns to two decimal places where appropriate
    for col in ["Missing Values (%)"]:
        if col in summary.columns:
            summary[col] = summary[col].apply(lambda x: f"{x:.2f}" if pd.notna(x) else "N/A")

    display(summary)

# Apply the function to the specified DataFrames
summarize_dataframe(users, "users")
summarize_dataframe(ratings, "ratings")
summarize_dataframe(movies, "movies")
```

```python
# @title 성별로 영화평점 평균점수
mean_ratings = data.pivot_table("rating", index="title",
                                columns="gender", aggfunc="mean")
mean_ratings
```

```python
# @title 영화평점이 250점 이상인 영화에 대한 성별에 따른 평균 평점
# 1. 영화 제목별 평점 개수 계산
ratings_by_title_correct = data.groupby("title").size()

# 2. 평점 개수가 250개 이상인 영화 제목만 필터링
active_titles = ratings_by_title_correct.index[ratings_by_title_correct >= 250]

# 3. 250개 이상인 영화에 대한 성별 평균 평점 추출
mean_ratings_filtered = mean_ratings.loc[active_titles]
mean_ratings_filtered
```

```python
# @title 특정 영화 제목 변경
mean_ratings = mean_ratings.rename(index={"Seven Samurai (The Magnificent Seven) (Shichinin no samurai) (1954)":
                           "Seven Samurai (Shichinin no samurai) (1954)"})

mean_ratings
```

```python
# 1. 250개 이상 평가를 받은 필터링된 데이터프레임 복사본 생성
mean_ratings_filtered = mean_ratings_filtered.copy()

# 2. 필터링된 데이터에 성별 평점 차이(diff) 열 추가
mean_ratings_filtered["diff"] = mean_ratings_filtered["M"] - mean_ratings_filtered["F"]

# 3. diff 기준으로 정렬
sorted_by_diff = mean_ratings_filtered.sort_values("diff")

# 4. 역순(남성이 여성보다 더 높게 평가한 순)으로 상위 5개 출력
sorted_by_diff[::-1].head()
```
