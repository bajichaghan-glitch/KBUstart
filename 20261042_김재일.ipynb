```python
import os
import requests
import sqlite3

DB_URL = "https://github.com/lerocha/chinook-database/raw/master/ChinookDatabase/DataSources/Chinook_Sqlite.sqlite"
DB_PATH = "chinook.db"

# 1. 기존의 잘못된(빈) 파일이 있다면 삭제합니다.
if os.path.exists(DB_PATH):
    os.remove(DB_PATH)
    print("🧹 Existing empty database removed.")

# 2. 데이터베이스를 새로 다운로드합니다.
print("📥 Downloading Chinook database...")
response = requests.get(DB_URL)
with open(DB_PATH, "wb") as f:
    f.write(response.content)
print("✅ Download complete.")

# 3. 다시 테이블 목록을 조회하여 확인합니다.
conn = sqlite3.connect(DB_PATH)
cursor = conn.cursor()
cursor.execute("SELECT name FROM sqlite_master WHERE type='table';")
tables = cursor.fetchall()
print("📊 Available tables in the database:", [table[0] for table in tables])
conn.close()
```

```python
# @title 모든 테이블을 사전형으로 읽기
# Initialize a dictionary to store all DataFrames
dfs = {}

# Connect to the SQLite database using a 'with' statement
with sqlite3.connect(DB_PATH) as conn:
    cursor = conn.cursor()
    cursor.execute("SELECT name FROM sqlite_master WHERE type='table';")
    tables = cursor.fetchall()

    # Iterate through each table and load it into a DataFrame
    for table_name in [table[0] for table in tables]:
        query = f"SELECT * FROM {table_name}"
        df = pd.read_sql_query(query, conn)
        dfs[table_name] = df
        print(f"--- {table_name} DataFrame head ---")
        display(df.head())
        print("\n")

print("All tables loaded into 'dfs' dictionary.")
```

```python
def summarize_df(df: pd.DataFrame) -> pd.DataFrame:
    # 데이터프레임의 각 컬럼별 데이터 타입, 결측치 수/비율, 대표값 5개를 요약한 DataFrame을 반환합니다.
    summary_data = []
    for col in df.columns:
        dtype = df[col].dtype
        n_unique = df[col].nunique()
        null_cnt = df[col].isnull().sum()
        null_ratio = round((null_cnt / len(df)) * 100, 2)
        unique_vals = df[col].dropna().unique()
        sample_vals = list(unique_vals[:7])

        summary_data.append({
            'Column': col,
            'N_Unique': n_unique,
            'Dtype': dtype,
            'Null_Count': null_cnt,
            'Null_Ratio(%)': null_ratio,
            'Sample_Values_5': sample_vals
        })

    return pd.DataFrame(summary_data)
```

```python
output_excel_path = 'all_tables_summary.xlsx'

# openpyxl 엔진을 사용하여 Excel 파일을 생성합니다. (Colab 기본 탑재)
with pd.ExcelWriter(output_excel_path, engine='openpyxl') as writer:
    for table_name, df in dfs.items():
        print(f"Summarizing {table_name}...")
        summary_df = summarize_df(df)
        # Write each summary DataFrame to a separate sheet
        summary_df.to_excel(writer, sheet_name=table_name, index=False)
        print(f"Summary for {table_name} written to sheet '{table_name}'.")

print(f"All table summaries saved to '{output_excel_path}'")
```

```python
# 0. PostgreSQL 연결에 필요한 드라이버 패키지 설치
!pip install -q psycopg2-binary

import sqlite3
import pandas as pd
from sqlalchemy import create_engine

# 1. Supabase PostgreSQL 연결 정보 설정
# ⚠️ 아래 정보를 본인의 실제 Supabase Database 정보로 변경하셔야 정상 작동합니다.
DB_USER = "postgres"
DB_PASSWORD = "your_supabase_password"  # 👈 여기에 실제 DB 비밀번호를 입력하세요.
DB_HOST = "db.xxxx.supabase.co"       # 👈 여기에 실제 Supabase 호스트 주소를 입력하세요.
DB_PORT = "5432"
DB_NAME = "postgres"

# SQLAlchemy 엔진 생성 (psycopg2 드라이버 명시)
DATABASE_URL = f"postgresql+psycopg2://{DB_USER}:{DB_PASSWORD}@{DB_HOST}:{DB_PORT}/{DB_NAME}"
engine = create_engine(DATABASE_URL)

# 2. with 구문으로 로컬 SQLite chinook.db 읽기
try:
    with sqlite3.connect('chinook.db') as sqlite_conn:
        # 모든 테이블 목록 조회
        table_names = pd.read_sql(
            "SELECT name FROM sqlite_master WHERE type='table';",
            sqlite_conn
        )['name'].tolist()

        # 3. 각 테이블을 읽어서 Supabase(PostgreSQL)로 전송
        for table in table_names:
            print(f"'{table}' 테이블 업로드 중...")

            # 로컬에서 데이터프레임으로 읽기
            df = pd.read_sql(f"SELECT * FROM {table}", sqlite_conn)

            # PostgreSQL 표준에 맞게 테이블 이름을 소문자로 변환
            target_table = table.lower()

            # Supabase로 데이터프레임 적재 (to_sql 활용)
            df.to_sql(
                name=target_table,
                con=engine,
                if_exists='replace',
                index=False
            )

    print("🎉 모든 Chinook 테이블이 Supabase로 성공적으로 마이그레이션되었습니다!")

except Exception as e:
    print("\n❌ 오류가 발생했습니다. DB_HOST와 DB_PASSWORD가 실제 값으로 정확히 변경되었는지 확인해 주세요.")
    print(f"상세 오류: {e}")
```

```python
import sqlite3
import pandas as pd

# SQLite 데이터베이스 연결
DB_PATH = 'chinook.db'

with sqlite3.connect(DB_PATH) as conn:
    # 고객별 총 구매 항목 수(Quantity) 및 총 구매 금액(Total) 집계 쿼리
    query = """
    SELECT
        c.CustomerId,
        c.FirstName || ' ' || c.LastName AS CustomerName,
        c.Country,
        SUM(il.Quantity) AS TotalTracksPurchased,
        ROUND(SUM(il.UnitPrice * il.Quantity), 2) AS TotalSpent
    FROM Customer c
    JOIN Invoice i ON c.CustomerId = i.CustomerId
    JOIN InvoiceLine il ON i.InvoiceId = il.InvoiceId
    GROUP BY c.CustomerId
    ORDER BY TotalTracksPurchased DESC, TotalSpent DESC
    LIMIT 5;
    """

    top_customers_df = pd.read_sql_query(query, conn)

display(top_customers_df)
```
